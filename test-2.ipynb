```python
#Webページ取得に使用するライブラリ
import requests

url =  "https://www.vegalta.co.jp/"

response = requests.get(url)

print("ステータスコード:", response.status_code)

print(response.text[:300])
```

```python
#HTMLを解析するライブラリ
from bs4 import BeautifulSoup

html = response.text

soup = BeautifulSoup(html, "html.parser")

soup
```

```python
#先のBeautiful Soupを利用して、タイトルを取得する
title = soup.title.text

print("タイトル", title)
```

```python
#先のBeautiful Soupを利用して、pタグを取得する
paragraphs = soup.find_all("p")

full_text = " " .join(p.get_text().strip() for p in paragraphs)

print(full_text[:300])
```

```python
#meta　description（ページの説明文）を探す
meta = soup.find("meta", attrs={"name":"description"})

description = meta["content"] if meta and meta.get("cotent") else ""

print("説明文:", description)
```

```python
#meta keywordを探す
meta_kw = soup.find("meta", attrs={"name":"keywords"})

if meta_kw and meta_kw.get("content"):
    keywords = [kw.strip() for kw in meta_kw["content"].split(",")][:10]
else:
    keywords = []

print("キーワード:", keywords )
```

```python
#HP内のリンクを保存するための空リスト
links = []

# href属性を持つaタグを全て探す
for a in soup.find_all("a", href=True):

    href = a["href"]

    # 外部リンク（httpから始まるもの）だけ保存
    if href.startswith("http"):
        links.append(href)

print("リンク数:", len(links))
print("リンク一覧（先頭3件）：" , links[:3])
```

```python
#ここまでやった、Webページ取得の部分を、fetch_page()にまとめる
from typing import Optional

def fetch_page(url: str, timeout: int = 10) -> Optional[str]:
     """
        指定URLのHTMLを取得する。
    
        Args:
            url: 取得対象URL
            timeout: タイムアウト秒数
    
        Returns:
            HTML文字列。失敗時は None
    """
     try:
          #User-Agentを指定してアクセス
          headers = {"User-Agent":"Tech0SearchBot/1.0(Educational Purpose)"}

          #Web ページにアクセス
          resp = requests.get(url, headers=headers, timeout=timeout)

          #エラーコードから例外を発生
          resp.raise_for_status()

          #文字コードを指定して設定
          resp.encoding = resp.apparent_encoding

          #HTML全文を返す
          return resp.text

     except requests.RequestException as e :
          #　取得失敗時は　Noneを返す
          print(f"✖️ 取得エラー: {e}")
          return None

```

```python
import re

def parse_html(html: str, url: str) -> dict:
   """
   HTMLを解析してページ情報を抽出する。
   
   Args:
      html: HTML文字列
      url: 元URL
   
   Returns:
      抽出した情報の辞書
   """ 

   #HTMLをBeautiful Soupで解析
   soup = BeautifulSoup(html,"html.parser")

   #不要タグを除去
   for tag in soup(["script","style","nav","footer","header"]):
      tag.decompose()

   #タイトル取得
   title = "No Title"
   if soup.find("title"):
      title = soup.find("title").get_text().strip()
   elif soup.find("h1"):
      title = soup.find("h1").get_text().strip()

   #meta descriptiom 
   description = ""
   meta = soup.find("meta", attrs={"name": "description"})
   if meta and meta.get("content"):
      description = meta["content"][:200]

   #meta keywords
   keywords = []
   meta_kw = soup.find("meta" , attrs={"name":"keywords"})
   if meta_kw and meta_kw.get("content"):
      keywords = [kw.strip() for kw in meta_kw["content"].split(",")][:10]

   #本文テキスト
   elems = soup.find_all(["p","h1","h2","h3","h4","h5","h6","li","td"])
   full_text = " ".join(e.get_text().strip() for e in elems)
   full_text = re.sub(r"\s+", " ",full_text).strip()

   #リンク
   links = [
      a["href"]
      for a in soup.find_all("a", href= True)
      if a ["href"].startswith("http")
   ][:20]

   return {
      "url":url,
      "title":title,
      "description":description,
      "keywords":keywords,
      "full_text":full_text,
      "links":links,
      "word_count": len(full_text.split()),
      "crawled_at":datetime.now().isoformat(),
      "crawl_status":"success" ,
    }
   
```

```python
#fetch_padeとparse_html を繋いで、crawl_urlを作ります。

def crawl_url(url: str ) -> dict:
    """
    URLをクロールして情報を返す（fetch → parse のワンストップ）。
    
    Args:
        url: クロール対象URL
    
    Returns:
        ページ情報の辞書（失敗時も crawl_status で判別可能）
    """
     # HTML取得
    html = fetch_page(url)

    # 取得失敗時
    if not html:
        return {
            "url": url ,
            "crawl_status" : "failed",
            "crawled_at": datetime.now().isoformat(),
            "error":"Failed to fetch page"
        }

    try:
        return parse_html(html, url)

    except Exception as e:
        return {
            "url":url,
            "crawl_status":"error",
            "crawled_at": datetime.now().isoformat(),
            "error":str(e),
        }
```

```python
result = crawl_url("https://www.apple.com/")

if result.get("crawl_status") == "success":
    print("✅ クロール成功!")
    print(f"📄 タイトル: {result['title']}")
    print(f"📝 説明: {result['description'][:100]}...")
    print(f"📊 文字数: {result['word_count']}語")
    print(f"🔗 リンク数: {len(result['links'])}件")
else:
    print(f"❌ クロール失敗: {result.get('error')}")
```
